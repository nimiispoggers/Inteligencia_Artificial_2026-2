##9 - Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.

```python
import numpy as np

##array com 10 números inteiros aleatórios de 1 a 100
array = np.random.randint(1, 100, 10)

#normalização
arrayNormalizado = (array - np.min(array)) / (np.max(array) - np.min(array))

print(f'''Array:
{array}
''')
print(f'''Array normalizado:
{arrayNormalizado}
''')
```

Para normalizar os valores entre 0 e 1, precisamos da "fórmula" dessa questão:

**normalizado = x - mínimo/ máximo - mínimo**

Sendo x = array;

mínimo = **np.min(array)**

máximo = **np.max(array)**

##10 - Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.

```python
import numpy as np

#matriz identidade 6x6
matrizIdentidade = np.eye(6)

#substituição da diagonal principal
for i in range(6):
  matrizIdentidade[i, i] = i + 1


print(f'''Matriz Identidade:
{matrizIdentidade}''')
```

Na **Apresentação da Biblioteca NumPy**, utilizei o exercício **6 - Geração de Matrizes Identidade** para geração de uma matriz identidade 6x6.

Ao fazer o comando **for i in range(6)**, pela matriz possuir seis linhas e seis colunas, o comando a ser executado em seguida, vulgo **matrizIdentidade[i, i] = i + 1**, vai passar por cada linha e coluna, e aumentar o número que está em +1.

Exemplo: se **for i in range(6)** estiver no 1° número, o comando abaixo será **matrizIdentidade[0, 0] = 0 + 1**, quando estiver no segundo número, será: **matrizIdentidade[1, 1] = 1 + 1**, deixando esse número no valor 2.

##1 - Calcule a integral da função f(x) = x^2 sin(x) no intervalo de 0 a π.


```python
import numpy as np
from scipy import integrate

#definir a função a ser integrada
def f(x):
  return x ** 2 * np.sin(x)

#calcular a integral unidimensional
resultado, erro = integrate.quad(f, 0, np.pi)

print(resultado)
print(erro)
```

##2 - Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.



```python
from scipy.integrate import solve_ivp

#função da equação diferencial
def f(x, y):
    return x - y

#intervalo de 0 até 10
intervalo = (0, 10)

#condição inicial y(0) = 1
y_inicial = [1]

#resolução da equação
solucao = solve_ivp(f, intervalo, y_inicial)

#exibe os valores
print("x:", solucao.t)
print("y:", solucao.y[0])
```

##4 - Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1

```python
import numpy as np
from scipy.optimize import minimize

#função a ser maximizada
def f(x):
  return -x ** 2 + 4 * x + 1

#condições de otimização
def fNegativo(x):
  return -f(x)

resultado = minimize(fNegativo, 0)

print(f'''Ponto Máximo (x): {resultado.x}
Valor Máximoo f(x): {f(resultado.x)}''')
```

Na **Apresentação da Biblioteca SciPy**, utilizei o exercício **4 - Otimização de Funções**. Para maximizar a função, retornamos ela com valor negativo, pois minimizando o negativo maximiza a função original.

##5 - Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
import numpy as np
from scipy.fft import fft

#sinal composto
def sinal(t):
  return np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

#tempo
t = np.linspace(0, 1, 1000)

#transformada de Fourier
transformada = fft(sinal(t))

print(transformada)
```

##10 - Calcule a integral dupla da função f(x, y) = xy + x^2 no intervalo [0, 2] para x e [0, 1] para y.

```python
import numpy as np
from scipy.integrate import dblquad

#função a ser integrada
def f(x, y):
  return x * y + x ** 2

#integral dupla
integral, erro = dblquad(f, 0, 2, lambda x: 0, lambda x: 1)

print(integral)
```

O **dblquad()** resolve as integrais duplas, fazendo com que os argumentos da função tenha a ordem alterada, O comando **lambda** exige que os limites das integrais sejam passadas para funções.

##12 - Verifique quais colunas de um DataFrame contêm valores duplicados.

```python
import pandas as pd

#criação do Dicionário
dados = {
    'Produto': ['Arroz', 'Feijão', 'Carne Moída', 'Café', 'Maionese'],
    'Preço': [10, 20, 30, 20, 10],
    'Quantidade': [1, 2, 3, 4, 5]
}

#criação do DataFrame
df = pd.DataFrame(dados)

for i in df.columns:
  if df[i].duplicated().any():
    print(f"A coluna '{i}' possui valores duplicados")
  else:
    print(f"A coluna '{i}' não possui valores duplicados")
```

Utilizei do **for i in df.columns:** para que o "i" percorresse por cada coluna do DataFrame, em seguida entra o **if df[i].duplicated()any():**, que verifica se naquela coluna que o "i" passou mais de uma vez.

##8 - Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).
