#10 - Sequência de Fibonacci

```python
#0 e 1 são os dois primeiros números dessa sequência
a=0
b=1
for i in range(10): #Repetirá o comando 10 vezes
  print(a)
  a, b = b, a + b #Fórmula de Fibonacci
```

**a = 0 e b = 1** são os valores básicos para iniciar essa fórmula/sequência

**for i in range()** para saber quantas vezes o comando irá se repetir, colocando o valor máximo que irá se repetir dentro dos parênteses, nesse caso, 10.

**a, b = b, a+b** como funciona essa conta? no lado esquerdo, a letra "a" se torna a letra "b" do lado direito, pois como na fórmula, a = b. Logo em seguida, a letra "b" do lado esquerdo se torna a soma de "a+b".

Exemplo: Imagine que se repetirá o comando até 3 vezes.
"a, b = b, a + b" se torna

"0, 1 = 1, 0 + 1" 1° volta

"1, 1 = 1, 1 + 1" 2° volta

"1, 2 = 2, 1 + 2" 3° volta

a = 1, então o programa vai imprimir esse "1"

#18 -  Número Primo
Crie um programa que verifique se um número fornecido é primo.

```python
numero = int(input("Digite um número para verificar se é primo: "))
if numero > 1:
  primo = True #consideramos o número como primo

  for i in range(2,numero): #perguntamos se o número é divisivel pelos números que vem antes dele
    if numero % i == 0: #se o número for divisivel por "i", ele não é primo
      primo = False #consideramos o número como não primo
      break

  if primo == True:
    print(f"{numero} é primo")
  else:
     print(f"{numero} não é primo")

elif numero == 0:
  print("Esse número é zero")
elif numero == 1:
  print("Esse número é um")
else:
  print("Esse número é negativo")
```

**primo = True** basicamente já considero que o número que o usuário digitou é primo. Mas caso ache um número que divida pelo número que o usuário digitou, ele retorna **primo = False**

**for i in range(2, numero):** "cria números" começando em 2 e indo até antes do número que o usuário digitou. Exemplo: se "numero" é igual a 5, o "range" vira "range(2, 5)" e verifica em seguida com o "if".

1° volta: 5/2; 2° volta/: 5/3; 3° volta: 5/4; 4° volta: 5/4. Parando sempre um número antes do i, se alguma dessas divisões tivesse a "sobra" == 0, dai não seria primo, pois teve algum número que dividia por ele mesmo.

**break** foi somente para parar o "for" caso já tenha achado algum divisor.

# 19 - Jogo de Adivinhação
Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

```python
import random
numero = random.randint(0,100) #a maquina vai escolher um número aleatório de 0 a 100
numeroUsuario = int(input("Digite um número entre 0 e 100: "))

while numeroUsuario != numero: #vai repetir o código até o usuário acertar o número
  if numeroUsuario > numero: #para o usuário ter mais chance de acertar o número se o número que ele escolheu for maior que a máquina escolheu
    print("Você errou, tente um número menor")
  else: #para o usuário ter mais chance de acertar o número se o número que ele escolheu for menor que a máquina escolheu
    print("Você errou, tente um número maior")
  numeroUsuario = int(input("Você errou, tente novamente: "))

print("Parabéns, você acertou")
```

**import random** foi utilizado para ser um gerador 100% aleatório, o comando "numero=random.randint(0,100)" faz com que a máquina escolha um número aleatório de 0 a 100.

**while** foi utilizado nessa situação foi caso o número que o usuário pediu for diferente do que a máquina "escolheu", ele vai avisar o usuário, dando uma dica também se o número é maior ou menor que o número que ele escolheu.

#21 - Busca Binária
Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
#a busca binária não verifica cada elemento. Ela começa pelo elemento no meio
lista = [10,20,30,40,50,60,70]
#Exemplo: o índice 0 corresponde ao núm 10, 1 ao 20, 2 ao 30, 3 ao 40 e etc.
numero = int(input("Digite o número que deseja procurar: "))
inicio = 0
fim = len(lista)-1 #nesse caso, 6
encontrado = False
while inicio <= fim:
  meio = (inicio+fim)//2 #nesse caso, índice 3, que corresponde ao número 40
  if lista[meio] == numero:
    encontrado = True
    break
  elif numero < lista[meio]: #se o usuário procurar por um número>40, o código pula esse "elif" e passa a procurar pela "parte direita do código"
    fim = meio -1
  else: #se o usuário procurar por um número<40, o código nem chega nesse "else", pois está procurando pela "parte esquerda" do código
    inicio = meio +1
if encontrado:
  print("Número achado!")
else:
  print("Número não achado!")
```

A busca binária faz com que a máquina comece a procurar o elemento no meio da lista. Foi inserido o "encontrado = false" pois ainda não achamos o elemento que o usuário quer.

**while** foi usado aqui para verificar onde está o número do usuário, se o número que o usuário digitou for == meio, ele já acha, caso não, ele verifica se o número que ele procura é maior ou menor que o meio, caso seja, ele vai para a direção correspondente.

#22 - Criptografia de Cifra de César
Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
def criptografar(mensagem, chave):
  alfabeto = "abcdefghijklmnopqrstuvwxyz" #criação do alfabeto
  mensagemCriptografada = ""
  mensagem = mensagem.lower()
  for i in mensagem: #"for" vai percorrer cada casa e verificar com o "if" se a letra está no alfabeto
    if i in alfabeto:
      local = alfabeto.index(i) #serve para descobrir a posição dela no alfabeto
      novo_local = (local + chave)%26 #"%26" serve para retornar ao alfabeto caso a chave faça a letra ultrapassar o limite de 26
      mensagemCriptografada += alfabeto[novo_local]
    else:
      mensagemCriptografada += i
  return mensagemCriptografada

mensagem = input("Digite a mensagem: ")
chave = int(input("Digite a chave: "))
resultado = criptografar(mensagem, chave)

print(f"Mensagem Criptografada: {resultado}")
```

**".lower()"** foi usado para deixar em minúsculo a frase do usuário.

**"for i in mensagem:
if i in alfabeto:
local = alfabeto.index(i)"** foram utilizados para descobrir no alfabeto se a letra existe no alfabeto e também para poder descobrir onde ela está localizada, o comando "novo_local = (local + chave)%26" serve para criptografar a mensagem, add a chave (contagem de caracteres a serem pulados para a criptografia) e o "%" serve para que, caso o valor passe dos 26 caracteres do alfabeto, ela retorne ao inicio.

#23 - Descriptografar Cifra de César

#24 - Validação de CPF
Implemente uma função para validar um CPF brasileiro.

```python
def validar(cpf):
  cpf=cpf.replace(".","").replace("-","") #caso o usuário digite "-" ou ".", esse comando vai retirá-la
  if len(cpf)!=11 or not cpf.isdigit(): #"len(cpf)" serve para ver se o tamanho do cpf está correto
  #o "cpf.isdigit()" serve para verificar se há somente números
    return False
  if cpf ==cpf[0] * 11: #evita que o cpf tenha todos os números iguais
    return False

  soma=0 #para adicionarmos depois os resultados das multiplicações
  for i in range(9): #vai passar pelos 9 primeiros números do CPF
    soma+=int(cpf[i])*(10-i) #o "int()" transforma a string em número
    #"10-i" pegaria os números de 10 até o número 2, depois faria cada um desses números multiplicar pelo número no cpf
    #Exemplo: cpf:523...: 10 x 5, 9 x 2, 8 x 3, etc, até ser 2 x 9°número do CPF

  restante=soma%11
  if restante<2: #se o resto da conta acima for = 0 ou 1
    primeiroDigito=0
  else: #caso contrário, será feita essa conta para verificar
    primeiroDigito=11-restante

  soma=0 #precisamos calcular o segundo dígito verificador, por isso zeramos a variável
  for i in range(10): #vai passar pelos 10 primeiros números do CPF
    soma+=int(cpf[i]) *(11-i) #o "int()" transforma a string em número

  restante=soma%11
  if restante<2: #se o resto da conta acima for = 0 ou 1
    segundoDigito=0
  else: #caso contrário, será feita essa conta para verificar
    segundoDigito=11-restante

  if int(cpf[9])==primeiroDigito and int(cpf[10])==segundoDigito:
    #o "primeiroDigito" serve para verificar se o penúltimo número está correto
    #o "segundoDigito" serve para verificar se o último número está correto
    return True #se ambos estiverem corretos, o cpf estará correto
  else:
    return False#caso contrário, o cpf estará incorreto
cpf=input("Digite seu CPF: ")
if validar(cpf):
  print("CPF válido!")
else:
  print("CPF inválido!")
```

#25 - Classificação de Números
Classifique uma lista de números em três categorias: positivos, negativos e zeros.

```python
lista = [10,1,31,-5,0,0,-6,-1,2]
positivo = [] #vai armazenar os números positivos
negativo = [] #vai armazenar os números negativos
zero = [] #vai armazenar os zeros
for i in lista: #para passar por cada número da lista
  if i > 0:
    positivo.append(i) #".append()" armazena o número à lista
  elif i < 0:
    negativo.append(i)
  else:
    zero.append(i)
print(f"Positivos: {positivo}")
print(f"Negativos: {negativo}")
print(f"Zeros: {zero}")
```

Foi criado três listas para armazenas cada um dos três tipos de números pedido pelo exercício.

#30 -Jogo da Forca
Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

```python
palavra="morango"
letrasDescobertas=["_"]*len(palavra)#lista com "_" para "esconder" as letras
#"*len(palavra)" serve para aparecer a quantidade de letras que tem a "palavra"
tentativas=5 #cada vez que o jogador errar, esse número diminui
letrasUsadas=[] #para guardar as letras já utilizadas
while tentativas>0 and "_" in letrasDescobertas: #o jogo continua enquanto ainda houver tentativas e ter "_" na palavra, mas faltando somente uma delas, o jogo acaba e o jogador perde
  print("Palavra:", " ".join(letrasDescobertas)) #mostra quais letras já foram mostradas
  print(f"Tentativas restantes: {tentativas}") #mostra quantos tentativas restam
  print(f"Letras usadas: {letrasUsadas}") #mostra quais letras já foram utilizadas

  letra=input("Digite uma letra: ").lower() #transforma a letra em minúscula
  if letra in letrasUsadas: #verifica se já usou aquela letra antes
    print("Você já tentou essa letra")
    continue #faz com que o programa volte para o "while" sem gastar uma das tentativas
  letrasUsadas.append(letra)
  if letra in palavra: #verifica se a letra está na palavra
    for i in range(len(palavra)): #o "for" vai passar até todas as letras, até o tamanho da palavra
      if palavra[i]==letra: #quando verificar todas, ele vê quais que correspondem, mudando o "_" para a letra colocada
        letrasDescobertas[i]=letra
    print("Certo")
  else:
    tentativas -=1 #comando para diminuir as tentativas
    print("Errado")
if "_" not in letrasDescobertas: #se não sobrar nenhum "_", quer dizer que o jogo acaba com vitória
  print(f"Parabéns, você descobriu a palavra: {palavra}")
else: #caso acabe as tentativas e ainda tem "_", o jogo acaba com derrota
  print(f"Você perdeu, a palavra era: {palavra}")
```

#31 - Ordenação por Bolha
Implemente o algoritmo de ordenação por bolha (Bubble Sort)

```python
lista=[5,9,6,5,2,1,]
for i in range(len(lista)): #controla quantas vezes vai passar pela lista
  for j in range(len(lista)-1-i): #percorre os elementos que ainda precisam ser ordenados
    if lista[j]>lista[j+1]: #se o número atual for maior que o próximo elemento, o valor é trocado
      lista[j], lista[j+1]=lista[j+1], lista[j]
#Exemplo: se a>b:
  #a,b=b,a
print(f"Lista em ordem: {lista}")
```

#35 - Números Fibonacci Até N

```python
numero=int(input("Digite o número que a sequência deve parar: "))
#0 e 1 são os dois primeiros números dessa sequência
a=0
b=1
while a <= numero:
  print(a)
  a,b=b,a+b #Fórmula de Fibonacci
```

**a, b = b, a+b** como funciona essa conta? no lado esquerdo, a letra "a" se torna a letra "b" do lado direito, pois como na fórmula, a = b. Logo em seguida, a letra "b" do lado esquerdo se torna a soma de "a+b".

Exemplo: Imagine que se repetirá o comando até 3 vezes.
"a, b = b, a + b" se torna

"0, 1 = 1, 0 + 1" 1° volta

"1, 1 = 1, 1 + 1" 2° volta

"1, 2 = 2, 1 + 2" 3° volta

a = 1, então o programa vai imprimir esse "1"

#50 - Cálculo de Frequência de Letras
Calcule a frequência de cada letra em uma string fornecida pelo usuário.

```python
palavra=input("Digite alguma palavra: ")
palavraFinal = palavra.replace(" ", "").lower()
frequencia={}

for i in palavraFinal:
  frequencia[i] = frequencia.get(i, 0) + 1

total=len(palavraFinal)

for i in frequencia:
  frequenciaRelativa = frequencia[i]/total
  print(f"{i} = {frequencia[i]} vezes que aparece = {frequenciaRelativa * 100:.2f}%")
```

No geral, essa questão é parecida com a questão 12 - Contagem de Caracteres e também foi utilizado elementos da questão 8 - Verificação de Palíndromo, como o ".replace(" ","")" e o ".lower()"

**total = len(palavra final))** foi utilizado pois é necessário saber a quantidade total de letras que foram contadas, para poder calcular a frequência.

**replace(" ","")**: Caso a string seja uma frase, ele retira essa parte vazia do texto, não contabilizando ao imprimir. **.lower()** foi a mesma coisa, para contabilizar a todas as letras, não separando por maiúscula ou minúscula.
